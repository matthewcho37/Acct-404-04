# Visualization using MatPlotLib and Seaborn

```python
# Total Units by Type into a Bar Chart 
total=data2026.groupby(['Type'])['Total Bulk and Bags'].sum()

# plot bar chart
total.plot(kind="bar",figsize=(10,6))

# print values
print(total)
plt.show()
```

```python
# Filter out data
ConventionalOnly=data2026[data2026['Type']!="Organic"]
```

```python
ConventionalOnly.describe()
```

```python
# Add a revenue column which calculates ASP times Total Bulk and Bags
data2026['Revenue']=data2026['ASP Current Year']*data2026['Total Bulk and Bags']
```

```python
data2026.head()
```

```python
# Bar Chart showing Revenue by Geography

revenue_by_geography=data2026.groupby(['Geography'])['Revenue'].sum()

#bar chart
revenue_by_geography.plot(kind="bar", figsize=(25,8))
```

```python
# Filter out total data by region
subtotals_to_exclude=['Total U.S.','West','Southeast','Northeast']
revenue_by_geography=data2026[~data2026['Geography'].isin(subtotals_to_exclude)] 
```

```python
# Bar Chart showing Revenue by Geography

revenue_by_geography1=revenue_by_geography.groupby(['Geography'])['Revenue'].sum()

#bar chart
revenue_by_geography1.plot(kind="bar", figsize=(25,8))
```

```python
# add sort to the chart
revenue_by_geography1_sorted=revenue_by_geography1.sort_values(ascending=False).head(20)
```

```python
# bar chart sorted
revenue_by_geography1_sorted.plot(kind='bar', figsize=(25,8)) 
```

```python
# Histogram on Average Selling Price

plt.figure(figsize=(10,10))
numeric_values=['ASP Current Year']
data2026[numeric_values].hist(bins=20,color="red",figsize=(20,10))
plt.show()
```

```python
# insert pie chart
labels='4225','4770','4046'
segments=data2026[['4225','4770','4046']].sum()
explode=(0,0,0)

fig1,ax1=plt.subplots()
ax1.pie(segments,explode,labels,shadow=True)
ax1.axis('equal') #equal means a circle

plt.show()
```
